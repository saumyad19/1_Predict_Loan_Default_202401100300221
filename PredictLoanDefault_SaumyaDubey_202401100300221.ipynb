```python
# Step 1: Upload the file in Google Colab
from google.colab import files
uploaded = files.upload()

# Step 2: Import pandas for data handling
import pandas as pd

# Step 3: Load the uploaded CSV file
# The uploaded file will be in the current working directory
file_name = list(uploaded.keys())[0]  # Gets the uploaded filename
df = pd.read_csv(file_name)  # Reads the CSV into a DataFrame

# Step 4: Display basic information about the dataset

# Display shape of the dataset
print("Dataset Shape:", df.shape)

# Display the column names
print("\nColumn Names:")
print(df.columns.tolist())

# Show information like column types and non-null counts
print("\nDataset Info:")
df.info()

# Show the first few rows of the dataset
print("\nFirst 5 Rows:")
print(df.head())
```
